기존 .env의 DATABASE_URL을 사용합니다. 자격 증명은 결과에 포함하지 않습니다.

```python
from pathlib import Path
import importlib.util
root = Path.cwd()
if not (root / "scripts").exists(): root = root.parent
import os; os.chdir(root)
spec = importlib.util.spec_from_file_location("monthly", root / "scripts/export_monthly_statistics.py")
monthly = importlib.util.module_from_spec(spec)
spec.loader.exec_module(monthly)
codes = [s.strip() for s in (root / "config/sgg_codes_nationwide.txt").read_text().splitlines() if s.strip() and not s.startswith("#")]
result = monthly.reconstruct_months(monthly.read_runs(), codes)
[(m["month"], m["polygon_count"], m["facility_point_count"], m["coverage_complete"]) for m in result["months"]]
```
